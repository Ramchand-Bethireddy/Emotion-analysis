```python
import numpy as np
import pandas as pd
import seaborn as sns
```

creating a dataframe

```python

df = pd.read_csv("/content/Restaurant_Reviews.tsv", sep='\t')
df
#https://www.kaggle.com/datasets/d4rklucif3r/restaurant-reviews    file to download before running
```

processing the data and data visualisation through seaborn

```python
y = df['Liked'].values
x = df['Review'].values
y
```

values for liked is 1

```python
sns.countplot(x ='Liked', data = df)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, random_state = 0)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vect = CountVectorizer(stop_words = 'english')
x_train_vect = vect.fit_transform(x_train)
x_test_vect = vect.transform(x_test)
```

```python
from sklearn.svm import SVC
model = SVC()
model.fit(x_train_vect, y_train)
```

```python
y_pred = model.predict(x_test_vect)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_pred, y_test)
```

```python
from sklearn.naive_bayes import MultinomialNB

from sklearn import metrics

acc = MultinomialNB().fit(x_train_vect, y_train)

y_pred= acc.predict(x_test_vect)

print("MultinomialNB Accuracy:",metrics.accuracy_score(y_pred, y_test ))
```

```python
import seaborn as sns
sns.distplot(y_pred)
```

```python
from sklearn.pipeline import make_pipeline
text_model = make_pipeline(CountVectorizer(),SVC())

text_model.fit(x_train, y_train)
y_pred = text_model.predict(x_test)
y_pred
```

```python
import joblib
joblib.dump(text_model, 'sentiment10')
```
